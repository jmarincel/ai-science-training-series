```python
def lookup_on_arxiv(query: str) -> str:
    """Find papers on the arxiv matching search criteria. 

    Parameters
    ----------
    query : str
        The search query. 

    Returns
    -------
    str
        A list of URLs pointing to papers on the arxiv. 
    """
    import arxiv
    
    # Construct the default API client.
    client = arxiv.Client()
    
    # Search for the 3 most recent articles matching the query.
    search = arxiv.Search(
        query=query,
        max_results=3,
        sort_by=arxiv.SortCriterion.Relevance
    )
    
    # `results` is a list. 
    results = [(r.title, r.entry_id) for r in client.results(search)]

    return results
```

```python
def download_arxiv_paper(arxiv_entry: tuple) -> int: 
    """Download pdf file of a paper from the arxiv. 

    Parameters
    ----------
    arxiv_entry : tuple
        A (title, URL) pair corresponding to a paper on the arxiv. 

    Returns
    -------
    null
    """
    import requests

    title, hyperlink = arxiv_entry
    pdf_link = hyperlink.replace('abs', 'pdf')
    filename = title + '.pdf'
    response = requests.get(pdf_link)
    with open(filename, mode="wb") as file:
        file.write(response.content)
    return 0
```

```python
lookup_on_arxiv("black to white hole transition")
```

```python
download_arxiv_paper(('Models for the nonsingular transition of an evaporating black hole into a white hole',
  'http://arxiv.org/abs/1811.06683v2'))
```
